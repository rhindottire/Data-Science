# KMeans Clustering

```python
import pandas as pd, matplotlib.pyplot as plt

from sklearn.cluster import KMeans
```

```python
data = {
    'X1': [2, 4, 4, 4, 6, 6],
    'X2': [4, 2, 4, 6, 4, 6]
}
df = pd.DataFrame(data)
```

```python
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
kmeans.fit(df)
```

**1. `kmeans = KMeans(...)` (Tahap Konfigurasi)**
Ini adalah langkah menyiapkan "alat" atau modelnya, tapi belum dijalankan:

* `n_clusters=2`: Kita memerintahkan algoritma untuk membagi data menjadi **2 kelompok**.
* `random_state=42`: Ini adalah "kunci" pengacak. Fungsinya agar setiap kali kode ini dijalankan, hasilnya **tetap sama/konsisten**.
* `n_init=10`: Algoritma akan mencoba melakukan pengelompokan sebanyak **10 kali** dengan posisi awal yang berbeda, lalu mengambil hasil yang paling bagus (paling rapi).

**2. `kmeans.fit(df)` (Tahap Eksekusi)**
Ini adalah perintah **"Kerjakan!"**.

* Fungsi `.fit()` menyuruh komputer untuk mempelajari data (`df`), menghitung jarak, dan melakukan pengelompokan sesuai konfigurasi di atas.

```python
labels = kmeans.labels_
centroids = kmeans.cluster_centers_

df['Cluster'] = labels + 1

print("Posisi Centroid Akhir:")
print(centroids)
```

```python
print("Hasil Pengelompokan:")
display(df)
```

```python
plt.figure(figsize=(6, 6))
colors = ['blue', 'green']

for i in range(2):
    cluster_data = df[df['Cluster'] == (i + 1)]
    plt.scatter(cluster_data['X1'], cluster_data['X2'],
                c=colors[i], label=f'Kelompok {i+1}', s=100)

plt.scatter(centroids[:, 0], centroids[:, 1], c='red', marker='X', s=200, label='Centroid')
plt.title('Hasil K-Means Clustering')
plt.xlabel('X1')
plt.ylabel('X2')
plt.legend()
plt.grid(True)
plt.show()
```
