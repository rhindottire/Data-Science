## Python Script for ChatGPT

````python
import openai

# Load your OpenAI API key
openai.api_key = "your-api-key-here"

def generate_chatgpt_response(prompt):
    response = openai.Completion.create(
        engine="text-davinci-003",  # Use the desired GPT model
        prompt=prompt,
        max_tokens=150,
        n=1,
        stop=None,
        temperature=0.7,
    )
    return response.choices[0].text.strip()

# Usage example
if __name__ == "__main__":
    user_input = input("Ask ChatGPT: ")
    chatgpt_response = generate_chatgpt_response(user_input)
    print(f"ChatGPT says: {chatgpt_response}")
````

## [Chat-GPT in Python](https://tomstechacademy.com/code-example-easiest-way-to-use-chat-gpt-in-python/)
Tutorial here : [YouTube](https://www.youtube.com/watch?v=Hsr779_pj9U)

````python
import os
from dotenv import load_dotenv
from openai import OpenAI
from openai import RateLimitError, AuthenticationError, OpenAIError

load_dotenv()

def get_api_key() -> str:
    api_key = os.getenv("CHATGPT_API_KEY")
    if not api_key:
        raise RuntimeError("CHATGPT_API_KEY tidak ditemukan di environment")
    return api_key
````

````python
client = OpenAI(api_key=get_api_key())
models = client.models.list()
model_ids = sorted([model.id for model in models.data])

model_ids
chat_models = [
    m for m in model_ids
    if m.startswith("gpt-")
]

chat_models
````

````python
def can_use_chat_model(model: str) -> bool:
    try:
        client.chat.completions.create(
            model=model,
            messages=[{"role": "user", "content": "ping"}],
            max_tokens=1,
        )
        return True
    except Exception:
        return False

chat_model_status = {
    model: can_use_chat_model(model)
    for model in chat_models
}

chat_model_status
````

````python
AVAILABLE_CHAT_MODELS = {
    model for model, ok in chat_model_status.items() if ok
}

def ask_openai(prompt: str, model: str = "gpt-4o-mini") -> str:
    if model not in AVAILABLE_CHAT_MODELS:
        raise ValueError(f"Model '{model}' tidak tersedia untuk API key ini")

    response = client.chat.completions.create(
        model=model,
        messages=[{"role": "user", "content": prompt}],
    )

    return response.choices[0].message.content
````

````python
def translate_openai_error(err: Exception) -> str:
    if isinstance(err, RateLimitError):
        return (
            "OpenAI API quota habis atau billing belum aktif.\n"
            "https://platform.openai.com/settings/organization/billing/overview"
        )

    if isinstance(err, AuthenticationError):
        return "API key OpenAI tidak valid."

    if isinstance(err, OpenAIError):
        return f"OpenAI API error: {err}"

    return f"Unexpected error: {err}"
````

````python
prompt = "Whats the most popular ski resort in Europe?"

try:
    result = ask_openai(prompt)
    print(result)

except Exception as e:
    print(translate_openai_error(e))
````
